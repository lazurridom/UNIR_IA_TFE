# DescargaDAICWOZ

El objetivo de este código es descargar y descomprimir de forma automática las particiones del dataset DAIC-WOZ, alojados en una URL

## 0. Setup
---

```python
import time
import zipfile
import requests
from pathlib import Path
from urllib.parse import urljoin, urlparse
from pathlib import Path
from bs4 import BeautifulSoup
```

```python
# Configuración URL DAIC-WOZ
url_daicwoz = "https://dcapswoz.ict.usc.edu/wwwdaicwoz/"
```

## 1. Funciones
---

```python
# Obtiene los enlaces de los archivos descargables de la web DAIC-WOZ
def get_links(url):

    t0 = time.perf_counter()

    r = requests.get(url, timeout=30)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")
    links = []
    for a in soup.find_all("a", href=True):
        href = a["href"]
        if href.startswith("?") or href.startswith("/") or href == "../":
            continue
        links.append(urljoin(url, href))

    elapsed = time.perf_counter() - t0
    print(f"{len(links)} archivos encontrados en la web. ({elapsed:.2f}s)")
    return links


# Dados los enlaces anteriores, descarga sus archivos en local
def download_files(links, out) -> None:

    t0 = time.perf_counter()
    out.mkdir(exist_ok=True)

    for url in links:
        name = Path(urlparse(url).path).name
        dest = out / name

        try:
            r = requests.get(url, stream=True, timeout=60)
            r.raise_for_status()
            with open(dest, "wb") as f:
                for chunk in r.iter_content(8192):
                    f.write(chunk)
            print(f"[OK] {name}")
        except Exception as e:
            print(f"[ERROR] {name}: {e}")

    print(f"\nTotal: {time.perf_counter() - t0:.2f}s")


# Extrae los zips de la ruta local especificada
def extract_zip(folder, delete=True):
    
    t0 = time.perf_counter()

    zips = list(Path(folder).glob("*.zip"))
    print(f"{len(zips)} ZIPs encontrados")

    for zip_path in zips:
        out_dir = zip_path.parent / zip_path.stem
        out_dir.mkdir(exist_ok=True)
        with zipfile.ZipFile(zip_path, "r") as zf:
            zf.extractall(out_dir)
        print(f"[OK] {zip_path.name}")
        if delete:
            zip_path.unlink()

    print(f"\nTotal: {time.perf_counter() - t0:.2f}s")
```

## 2. Ejecución
---

```python
links = get_links(url_daicwoz)
```

```python
download_files(
    links
    , Path(r'D:\DAIZ-WOZ\base')
)
```

```python
extract_zip(
   Path(r'D:\DAIZ-WOZ\base')
)
```
